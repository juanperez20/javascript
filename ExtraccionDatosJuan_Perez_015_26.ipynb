##Receta 2: Extraccion de Textos de Documentos PDF
los archivos de PDF son comunes para documnetos formales. Extraer textos de ellos requieren bibliotecas especificas que puedan interpretar su estructura.

````python
#Generar un archivo PDF de prueba con el texto de relleno usado ReportLab
from reportlab.lib.pagesizes import letter
from reportlab.pdfgen import canvas

nombre_archivo_pdf = "JuanPerez_20260929.pdf"
contenido_pdf = "Este es mi PDF de prueba + mi texto de relleno personalizado" + texto_relleno

c = canvas.Canvas(nombre_archivo_pdf, pagesize=letter)
c.setFont("Helvetica", 12)
c.drawString(72, 750, "Ejemplo de Documento PDF para la Extraccion de Datos")

# Dividir el contenido para ajustar en la pagina
y_position = 730
line_height = 14
for line in contenido_pdf.split('\n'):
  c.drawString(72, y_position, line)
  y_position -= 20
  if y_position < 50:
    c.showPage()
    y_position = 750
    c.setFont("Helvetica", 12)

c.save()

print(f"Archivo PDF creado: {nombre_archivo_pdf} con exito.")
````

````python
#Extraer texto de un PDF

import PyPDF2

texto_PDF_extraido = ""

try:
  with open(nombre_archivo_pdf, "rb") as archivo_pdf:
    lector_pdf = PyPDF2.PdfReader(archivo_pdf)
    for page in lector_pdf.pages:
      texto_pdf_extraido += page.extract_text() + "\n"

except Exception as e:
  print(f"Error al extraer texto del PDF: {e}")
  texto_pdf_extraido = "Error al extraer texto de PDF."

print("\n--- Contenido extraido (primeros 300 caracteres) ---")
print(texto_PDF_extraido[:300] + "...")
print("\nLongitud total del texto extraido:", len(texto_pdf_extraido), "caracteres.")
````

##Receta 4 : Extraccion de texto de Paginas Web (Web Scraping)

el Web Scraping nos permite extraer informacion directamete de sitios web. Usaremos ``requests`` para obtener el HTML y ``BeautifulSoup`` para parsearlo.

###Entendiendo la estructura Basica de HTML para Web Scraping

Para extraer información de una página web, primero necesitamos entender cómo está estructurada. HTML (HyperText Markup Language) es el lenguaje estándar para crear páginas web y está compuesto por una serie de elementos o ``etiquetas`` anidadas que definen el contenido y su organización.   
Consideremos la página de ejemplo ``http://quotes.toscrape.com/``. Si inspeccionas está página (usando las herramientas de desarrollo de tu navegador ``F12``) verás una estructura similar a la siguiente:

```html
<div class="quote">
 <span class="text" itemprop="text">"La vida es lo que pasa mientras estás ocupado haciendo otros planes."</span>
 <span>
  by <small class="author" itemprop = "author"> John Lennon</small>
  <a href="/author/John-Lennon">(más)</a>
 </span>
 <div class="tags">
  Etiquetas: <a class="tag" href="/tag/life/" (...) >life</a>
  <a class="tag" href="/tag/live" (...) >live</a>
 </div>
</div>
```

**Elementos claves**

* **Etiquetas (``<div>``, ``<span>``, ``<small>``, ``<a>``):** Indica el tipo de contenido. Por ejemplo, ``<div>`` suele ser contenedores, ``<span>`` para texto en línea, ``<small>`` para texto más pequeño y ``<a>`` para enlaces.

  * **Atributos (``class``, ``href``, ``itemprop``):** Proporcionann información adicional sobre etiquetas. Son cruciales para seleccionar elementos específicos:
  * ``class="quote"``: Indica que este ``div`` es un bloque de cita.
  * ``itemprop="text"``: Define el texto de la cita.
  * ``class="author"``: Identifica el autor.
  * ``href="/author/John-Lennon"``: Contiene la URL del perfil del autor.

Cuando hacemos web scraping nuestro objetivo es localizar estas etiquetas y atributos específicos para estrar la información deseada

````python
import requests
from bs4 import BeautifulSoup

url_ejemplo = "http://quotes.toscrape.com/" # Una página diseñada para practicar Web Scraping

texto_web_extraido = ""

try:
    response = requests.get(url_ejemplo)
    response.raise_for_status() # Lanza una excepción para errores HTTP
    soup = BeautifulSoup(response.text, "html.parser")

    # Extraer el texto de todas las citas y autores
    citas = soup.find_all("span", class_="text")
    autores = soup.find_all("small", class_="author")

    for i in range(len(citas)):
      texto_web_extraido += f"Cita: {citas[i].get_text()}\n"
      texto_web_extraido += f"Autor: {autores[i].get_text()}\n\n"
except requests.exceptions.RequestException as e:
    print(f"Error al acceder a la página web: {e}")
    texto_web_extraido = "Error al acceder a la página web."
except Exception as e:
    print(f"Error al extraer el texto de la página web: {e}")
    texto_web_extraido = "Error al extraer texto de la página web."

print("\n--- Contenido extraído (primeros 300 caracteres) ---")
print(texto_web_extraido[:300] + "...")
print("\nLongitud total del texto extraído de la web:", len(texto_web_extraido), "carecteres.")
````

Admás de HTML las páginas web utilizan CSS para definir su apariencia y fomato (colores, fuentes, diseños y etc). Aunque HTML estructura el contenido CSS le da estilo. Los selectores CSS (como ``class="text"`` o ``id="main-content"``) son fundamentales en Web Scraping ya que nos permiten identificar y seleccionar elementos específicos basándonos en cómo están estilizados o agrupados, facilitando la extracción precisa de los datos.

**Nota Importante sobre Páginas Dinámicas:** Es crucial entender que ``requests`` y ``BeautifulSoup`` son excelentes para páginas web estáticas donde el contenido HTML ya está presente en la respuesta inicial del servidor. Sin embargo muchas páginas modernas cargan su contenido de forma dinámica utilizando JavaScript. En estos casos ``requests`` solo verá el HTML inicial (a menudo vacío o con plantillas) y ``BeautifulSoup`` no podrá extraer el contenido final. Para estas situaciones, herramientas como **Selenium** (un software para la automatización de navegadores web) son necesarias ya que simulan la interacción de un usuario real en un navegador permitiendo que el JavaScript se ejecute y el contenido dinámico se cargue antes de la exracción. Por ahora no introduciremos ``Selenium`` en el código pero es fundamental conocer esta limitación ya su posible solución.

**Consideraciones Éticas:** Es fundamental respetar lo ``robots.txt`` de los sitios web y sus términos de servicio. Para este ejemplo usamos una página pública de ejemplo.

## Receta 5: Extraccion de Datos de APIs
Las APIs (Interfaces de Programación de Aplicaciones) son una forma estructurada de acceder a los datos. Usaremos una API pública para demostra la extracción de datos JSON.

**Operaciones Comunes de APIs (Verbos HTTP):** Las APIs suelen seguir el protocolo HTTP y utilizan "verbos" o métodos para indicar el tipo de operación que se desea realizar sobre un recurso:

* **GET:** Se utiliza para solicitar datos de un recurso específico (Ejemplo: obtener información de un post).
* **POST:** Se utiliza para enviar datos y crear un nuevo recurso. (Ejemplo: crear un nuevo usuario).
* **PUT:** Se utiliza para enviar datos y actualizar un recurso existente, reemplazándolo completamente.
* **DELETE:** Se utiliza para eliminar un recurso específico.
* **PATCH:** Se utiliza para aplicar modificaciones parciales a un recurso.

**Estructura Básica de un Documento JSON:** JSON (JavaScript Object Notation) es un formato ligero para el intercambio de datos. Es fácil de leer y escribir para los humanos y fácil de parcear ya generar para las máquinas. Su estructura se basa en dos tipos:

* **Colecciones de pares nombre/valor (objetos):** Representados por corchetes ``{}``. Un objeto es una lista no ordenada de pares ``"clave": valor``. Las claves deben ser cadenas de texto.

```json
{
"nombre" : "Juan",
"edad" : 30,
"ciudad" : "Bogotá"
}
```

* **Listas ordenadas de valores (arrays):** Representadas por llaves ``[]``. Un array es una secuencia de valores.

```json
[
 "manzana",
 "banano",
 "limón"
]
```
Los valores pueden ser cadenas de texto, números, booleanos (``true``/``false``), null, objetos o arrays. Esta flexibilidad permite representar estructuras de datos complejas.   Para este ejemplo utilizaremos la API pública de
``JSONPlaceholder`` que proporciona datos ficticios de prueba.

````python
import requests
import json

url_api = "https://jsonplaceholder.typicode.com/posts/1" # Un post de ejemplo

datos_api_extraidos = {}

try:
    response = requests.get(url_api)
    response.raise_for_status()
    datos_api_extraidos = response.json()

except requests.exceptions.RequestException as e:
    print(f"Error al acceder a la API: {e}")
    datos_api_extraidos = {"Error": "Error al acceder a la API"}

except json.JSONDecodeError as e:
    print(f"Error al decodificar JSON de la API: {e}")
    datos_api_extraidos = {"Error": "Error al decodificar JSON"}

print("\n --- Datos Extraídos de la API ---")
print(json.dumps(datos_api_extraidos,indent=2))
print(f"\nTipo de datos extraídos: {type(datos_api_extraidos)}")
````

## Receta 6: Extracción de Datos de Bases de Datos

Las bases de datos (BD) permiten almacenar y organizar información para que pueda ser consultada, actualizada y utilizada posteriormente por diferentes aplicaciones.

En una base de datos relacional la información se organiza principalmente en tablas compuestas por filas y columnas. Una base de datos puede contener varias tablas relacionadas entre sí, por lo que es importante no confundir una tabla con la base de datos completa.

Para trabajar con bases de datos relacionales se utiliza principalmente ``SQL`` (Structured Query Language) un lenguaje que permite crear estructuras, almacenar información y realizar consultas sobre los datos.

En esta receta utilizaremos ``SQLite`` un sistema de base de datos relacional ligero que puede almacenar toda una base de datos en un único archivo. Python incorpora el módulo sqlite3 que nos permitirá crear la base de datos, almacenar algunos datos y posteriormente extraerlos.

### Crear nuestra base de datos

Antes de extraer información necesitamos una fuente que la contenga. Al igual que en las recetas anteriores creamos archivos .txt, .pdf o .docx, ahora crearemos un archivo que contendrá nuestra base de datos.

Utilizaremos ``sqlite3.connect()`` para conectarnos a **SQLite**. Si el archivo indicado todavía no existe, SQLite lo creará automáticamente.

````python
import sqlite3
import pandas as pd

# Nombre del archivo que contendrá nuestra base de datos
nombre_db = "AngieBlanco_20261006.db"
# Crear o conectarse a la base de datos
conn = sqlite3.connect(nombre_db)

print(f"Base de datos '{nombre_db}' creada/conectada con éxito")
````

### Crear una tabla para almacenar los datos

Nuestra base de datos ya existe pero todavía no contiene información. Necesitamos definir dónde almacenaremos los datos.

Crearemos una tabla llamada articulos con tres columnas:

* id: indentificador único de cada artículo.
* titulo: título del artículo.
* contenido: texto asociado al artículo

La instrucción **SQL CREATE TABLE** permite crear una nueva tabla y definir las columnas que formarán parte de ella.

````python
# Crear un cursos para ejecutar instrucciones SQL
cursor = conn.cursor()

# Crear la tabla articulos
cursor.execute('''
CREATE TABLE IF NOT EXISTS articulos (
 id INTEGER PRIMARY KEY,
 titulo TEXT NOT NULL,
 contenido TEXT
)
''')

# Guardar los cambios
conn.commit()

print("Tabla 'articulos' creada con éxito.")
````

### Insertar datos en la tabla

Ahora tenemos una base de datos y una tabla pero la tabla se encuentra vacía.

Crearemos algunos registros de prueba y utilizaremos la instrucción **SQL INSERT** para almacenarlos. En este ejemplo cada elemento de la lista representa una fila que será incorporada en la tabla articulos.

````python
# Datos de prueba
datos_prueba = [
    (1,
     "La importancia del Data Science",
     "El data science es crucial para la toma de decisiones modernas."),
    (2,
     "PNL: De la teoría a la práctica",
     "El procesamiento de lenguaje natural ha avanzado enormemente con el deep learning."),
    (3,
     "Web Scraping Ético",
     "Es fundamental seguir buenas prácticas al realizar web scraping.")
]

# Insertar los registros en la tabla
cursor.executemany(
    "INSERT OR IGNORE INTO articulos (id, titulo, contenido) VALUES (?, ?, ?)",
    datos_prueba
)

# Guardar los cambios realizados
conn.commit()

print(f"{len(datos_prueba)} registros procesados en la tabla 'articulos'.")

````

### Extraer datos mediante una consulta SQL

Ahora sí llegamos al objetivo principal de esta receta: extraer información almacenada en una base de datos.

Para consultar datos utilizaremos la instrucción **SQL SELECT**. Podemos indicar qué columnas queremos recuperar y de qué tabla deben obtenerse. Por ejemplo:

```sql
SELECT titulo, contenido
FROM articulos;

Esta consulta puede leerse prácticamente como una instrucción:

**"Seleccionar las columnas titulo y contenido de la tabla articulos"**

Ejectutaremos esta consulta desde Python y recuperaremos las filas encontradas

````python
# Ejecutar una consulta SQL
cursor.execute('''
SELECT titulo, contenido
FROM articulos
''')

# Recuperar todos los registros encontrados

filas = cursor.fetchall()

print("--- Datos Extraídos de la Base de Datos ---")

for fila in filas:
    print(f"Título: {fila[0]}")
    print(f"Contenido: {fila[1]}")
    print("---")

print(f"\n Nñumero de registros extraídos: {len(filas)}")
````

## Receta 8: Extracción de Datos de Archivos XML y JSON

**XML y JSON** son formatos comunes para el intercambio de datos estructurados. Python tiene módulos nativos para trabajar con ambos.

**XML** nos es un lenguaje con un vocabulario fijo, sino un conjunto de reglas para crear lenguajes de marcado. **SVG y RSS** son dos vocabularios especializados basados en **XML** pero resuelven problemas completamente diferentes.

**Relación entre tipos de estructuras**

| Tecnología | Relación | ¿Para qué sirve? |
|---|---|---|
| **HTML** | Lenguaje de marcado relacionado históricamente con SGML | Estructurar contenido de páginas web |
| **XML** | Lenguaje de marcado derivado de SGML | Representar e intercambiar datos estructurados |
| **SVG** | Lenguaje de gráficos basado en XML | Representar gráficos vectoriales |
| **RSS** | Formato de sindicación basado en XML | Distribuir contenido actualizado |
| **JSON** | Formato de intercambio de datos independiente de XML | Representar e intercambiar datos estructurados de forma ligera |

````python
import json
import xml.etree.ElementTree as ET

# Generar un archivo JSON de prueba
nombre_archivo_json = "JuanPerez_20260929.json"
datos_json = {
    "libro" :{
        "titulo" : "Aprendiendo Python",
        "autor" : "John Doe",
        "generos" : ["Programación", "Educación"],
        "disponible" : True
    }
}
with open(nombre_archivo_json, "w", encoding="utf-8") as archivo:
    json.dump(datos_json, archivo, indent=2)

print(f"Archivo '{nombre_archivo_json}' creado con éxito.")
````

````python
# Generar un archivo XML de prueba
nombre_archivo_xml = "JuanPerez_20260929.xml"
root = ET.Element("libros")
libro = ET.SubElement(root, "libro")
titulo = ET.SubElement(libro, "titulo")
titulo.text = "Data Science con Python"
autor = ET.SubElement(libro, "autor")
autor.text = "Jane Smith"
publicacion = ET.SubElement(libro, "publicacion")
publicacion.text = "2022"

tree = ET.ElementTree(root)
with open(nombre_archivo_xml, "wb") as f:
    tree.write(f, encoding="utf-8", xml_declaration=True)

print(f"Archivo '{nombre_archivo_xml}' creado con éxito.")
````

````python
# Extraer datos de JSON

datos_json_extraidos = {}
try:
    with open(nombre_archivo_json, "r", encoding="utf-8") as f:
        datos_json_extraidos = json.load(f)
except Exception as e:
    print(f"Error al leer el archivo JSON: {e}")
    datos_json_extraidos = {"Error": "Error al leer el archivo JSON"}

print("--- Datos Extraídos de JSON ---")
print(json.dumps(datos_json_extraidos, indent=2))
````

````python
# Extraer datos de XML
datos_xml_extraidos = []
try:
    tree = ET.parse(nombre_archivo_xml)
    root = tree.getroot()
    for libro in root.findall("libro"):
        titulo = libro.find("titulo").text
        autor = libro.find("autor").text
        publicacion = libro.find("publicacion").text
        datos_xml_extraidos.append({"titulo": titulo, "autor": autor, "publicacion": publicacion})
except Exception as e:
    print(f"Error al leer el archivo XML: {e}")
    datos_xml_extraidos =[{"error": "Error al leer XML"}]

print("\n --- Datos Extraídos de XML ---")
for item in datos_xml_extraidos:
    print(item)
````
